```python
import fastf1
import pandas as pd

for i in range(1, 8):  # 1'den 7'ye kadar olan ay sayıları
    race = fastf1.get_session(2024, i, 'R')
    race.load()
    laps = race.laps.pick_quicklaps()
    laps_df = pd.DataFrame(laps)
    file_name = f'2024_0{i}_quick.csv'  # Dosya adını oluştur
    laps_df.to_csv(file_name, index=False)

```
